# 🔄 Phase 3 (Alt 1): Setup Validation Cross-Project (90% Train Projects, 10% Test Projects)

**Penelitian**: *Predicting Flaky Test in Cross-Project Scenario (Alternative 1)*

> **[!IMPORTANT]**
> Pemisahan train-test dilakukan secara ketat pada level **proyek (project-level split)**, bukan random test split, untuk memastikan model diuji pada proyek *unseen* tanpa *data leakage*.

Notebook ini merealisasikan alternatif validasi:
- **Rasio Proyek**: **90% Proyek Pelatihan** (22 proyek) dan **10% Proyek Pengujian** (2 proyek).
- **Tujuan**: Menguji pengaruh memperbesar jumlah variasi proyek sumber (*source projects pool*) terhadap performa generalisasi model pada proyek target baru.

```python
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (12, 6)

data_path = "../data_alt_1/engineered_test_features.csv"
df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape[0]} baris, {df.shape[1]} kolom")
print(f"Total Unique Projects: {df['project'].nunique()}")

with open("../data_alt_1/validation_splits_metadata_90_10.json") as f:
    split_info = json.load(f)

print("=== Konfigurasi Split 90/10 ===")
print(f"Train Projects ({split_info['n_train_projects']}): {split_info['train_projects']}")
print(f"Test Projects  ({split_info['n_test_projects']}): {split_info['test_projects']}")
print(f"Train Samples : {split_info['train_samples']} (Flaky: {split_info['train_flaky']}, {split_info['train_flaky_pct']}%)")
print(f"Test Samples  : {split_info['test_samples']} (Flaky: {split_info['test_flaky']}, {split_info['test_flaky_pct']}%)")
```

```python
# Visualisasi Komposisi Train vs Test Projects
split_summary = pd.DataFrame([
    {'Set': 'Train (22 Proyek)', 'Samples': split_info['train_samples'], 'Flaky': split_info['train_flaky']},
    {'Set': 'Test (2 Proyek)', 'Samples': split_info['test_samples'], 'Flaky': split_info['test_flaky']}
])

plt.figure(figsize=(8, 5))
sns.barplot(data=split_summary, x='Set', y='Samples', palette='Blues_d')
plt.title('Distribusi Sampel pada Split 90% Train vs 10% Test Projects', fontweight='bold')
for i, row in split_summary.iterrows():
    plt.text(i, row['Samples'] + 200, f"{row['Samples']:,} samples\n(Flaky: {row['Flaky']})", ha='center', fontweight='bold')
plt.ylim(0, 22000)
plt.tight_layout()
plt.show()
```
